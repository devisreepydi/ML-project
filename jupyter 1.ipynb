```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the traffic dataset

file_path = "PeMSD7_V_228.csv"

# header=None because the CSV does not contain column names
data = pd.read_csv(file_path, header=None)

print("Dataset loaded successfully!")
```

```python
# Create meaningful column names for the 228 traffic sensors

data.columns = [f"Sensor_{i+1}" for i in range(data.shape[1])]

print("Column names assigned successfully!")
```

```python
# Dataset Exploration

# data.shape[0] gives the number of rows/observations
print("Number of observations:", data.shape[0])

# data.shape[1] gives the number of columns/attributes
print("Number of attributes:", data.shape[1])
```

```python
# Display the first 5 rows of the dataset

data.head()
```

```python
# Check for missing values in the dataset

# isnull() checks each cell for missing data
# sum().sum() gives the total number of missing values

print("Total missing values:", data.isnull().sum().sum())
```

```python
# Check for duplicate rows
print("Duplicate rows:", data.duplicated().sum())
```
